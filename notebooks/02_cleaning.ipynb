```python
import kaggle
import os
import numpy as np
import pandas as pd

path_get_raw_data = "../data/raw/"
processed_data_path = "../data/processed/"

raw_data1 = pd.read_csv(os.path.join(path_get_raw_data, "WA_Fn-UseC_-Telco-Customer-Churn.csv"))
```

```python
processed_data = raw_data1.copy()
processed_data.drop(columns=["customerID"], inplace=True)
processed_data.info()
```

```python
def summarize(df, max_show=20):
    for col in df.columns:
        s = df[col]
        n_unique = s.nunique()
        print(f"\n{'='*60}\n{col}  ({s.dtype}, {n_unique} unique)")
        if pd.api.types.is_numeric_dtype(s):
            print(f"  range : {s.min()} .. {s.max()}   mean {s.mean():.2f}")
        uniq = sorted(s.unique(), key=str)
        if n_unique <= max_show:
            print(f"  values: {uniq}")
        else:
            print(f"  sample: {uniq[:max_show]} ... (+{n_unique - max_show} more)")

summarize(processed_data)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

for col in processed_data.select_dtypes(include='object').columns:
    processed_data[col] = le.fit_transform(processed_data[col])
```

```python
processed_data.to_csv(processed_data_path + "processed_data.csv", index=False)
```
